## 1. Project Introduction

Antimicrobial resistance (AMR) is an important public health concern because it can make infections more difficult to treat and increase the risk of complications. Monitoring patterns of antibiotic resistance can help healthcare professionals and public health organisations understand the scale of the problem and identify areas that may require further investigation.

This project explores publicly available surveillance data relating to antibiotic resistance in *Escherichia coli* (*E. coli*) bloodstream infections in England. Using data analysis and visualisation, the project aims to examine the patterns and trends represented in the selected dataset and communicate relevant findings clearly.

The project will also consider the ethical responsibilities involved in working with healthcare-related data, including data governance, privacy, responsible interpretation, potential bias and the limitations of the available information.

The intended outcome is an evidence-based analysis supported by appropriate visualisations and a dashboard that communicates findings to a healthcare and public health audience.


## 2. Business Case and Problem Statement

### Business Case

Antimicrobial resistance (AMR) is a significant public health concern because it can reduce the effectiveness of medicines used to treat infections. Monitoring antibiotic resistance helps healthcare professionals and public health organisations understand resistance patterns and identify changes that may require further investigation.

Data analysis and visualisation will be used to explore the information provided by the dataset and communicate relevant findings.

### Problem Statement

The central problem addressed by this project is how to present antibiotic resistance surveillance data in a clear and meaningful way so that patterns and changes over time can be explored.

The project will investigate the available data, compare resistance measurements across the antibiotic groups represented in the dataset, and develop visualisations and a dashboard to communicate the findings.

The analysis will also consider data quality, reporting limitations, privacy, responsible interpretation and the ethical responsibilities associated with communicating healthcare-related information.
