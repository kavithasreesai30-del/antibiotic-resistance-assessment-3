### Summary Statistics by Antibiotic

This analysis will summarise antibiotic resistance measurements for each antibiotic category using descriptive statistics. Comparing the mean, median, minimum and maximum values will help identify differences in resistance levels across the seven antibiotic categories.


```python
# Calculate descriptive statistics for each antibiotic category
antibiotic_summary = df.groupby("stratum")["metric_value"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

# Display the summary statistics
display(antibiotic_summary)
```

### Initial Observations

Co-amoxiclav has the highest mean resistance value (42.69%), while carbapenems have the lowest (0.21%). Ciprofloxacin has a mean of 19.07%, followed by third-generation cephalosporins at 16.04%. All seven antibiotic categories contain 61 monthly observations.

These are descriptive findings from the dataset and do not establish why resistance levels differ. Further analysis will examine how resistance patterns change over time.


### Antibiotic Resistance Trends Over Time

A line chart will be used to examine monthly resistance patterns for all seven antibiotic categories from July 2021 to July 2026. This visualisation will help identify increases, decreases and fluctuations in resistance over time.


#### Code Explanation

The code sorts the data by date and plots a separate line for each antibiotic category. The chart includes a title, axis labels and a legend to make the results easier to read and compare.


```python
# Sort the data by date
df_sorted = df.sort_values("date")

# Create a line chart
plt.figure(figsize=(14, 7))

for antibiotic, group in df_sorted.groupby("stratum"):
    plt.plot(
        group["date"],
        group["metric_value"],
        label=antibiotic
    )

# Add chart labels and formatting
plt.title("Monthly E. coli Resistance Trends by Antibiotic in England")
plt.xlabel("Month")
plt.ylabel("Resistance (%)")
plt.legend(title="Antibiotic", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Interpretation of Resistance Trends

The line chart shows monthly antibiotic resistance patterns in *E. coli* bloodstream infections in England from July 2021 to July 2026. Co-amoxiclav consistently has the highest resistance percentages, while carbapenems have the lowest. Several antibiotic categories appear to show gradual increases over time, whereas others fluctuate without a clear overall trend. Amikacin and carbapenems remain relatively stable compared with the other categories. These observations describe patterns in the dataset and do not establish the causes of changes in resistance.


### Comparing Average Resistance by Antibiotic

A bar chart will compare the mean resistance percentage across the seven antibiotic categories. This will make it easier to compare resistance levels between antibiotics over the study period.


#### Code Explanation

The code calculates the mean resistance percentage for each antibiotic category and displays the results as a bar chart. The chart includes a title, axis labels and category names to make comparisons easier to understand.


```python
# Calculate the mean resistance percentage for each antibiotic
mean_resistance = df.groupby("stratum")["metric_value"].mean().sort_values()

# Create a bar chart
plt.figure(figsize=(10, 6))
mean_resistance.plot(kind="bar")

# Add chart labels and formatting
plt.title("Average E. coli Resistance by Antibiotic in England")
plt.xlabel("Antibiotic")
plt.ylabel("Mean Resistance (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

#### Interpretation of Average Resistance

The bar chart shows clear differences in mean antibiotic resistance percentages across the seven categories from July 2021 to July 2026. Co-amoxiclav has the highest mean resistance at 42.69%, while carbapenems have the lowest at 0.21%. Ciprofloxacin follows at 19.07%, and third-generation cephalosporins at 16.04%. Gentamicin and piperacillin with tazobactam have similar mean resistance levels, at 11.54% and 11.67%, respectively. Amikacin also shows a relatively low mean of 2.76%. These results describe differences in the dataset and do not explain their causes.


### Distribution of Resistance Percentages by Antibiotic

A box plot will be used to examine the distribution of resistance percentages for each antibiotic from July 2021 to July 2026. It will help compare the spread of the values and identify potential outliers across the seven antibiotic categories.


#### Code Explanation

The code groups the resistance percentages by antibiotic and displays their distributions using a box plot. The box plot compares the distribution of *E. coli* resistance percentages across seven antibiotic categories in England between July 2021 and July 2026. Co-amoxiclav has the highest resistance distribution, while carbapenems have the lowest. Ciprofloxacin and third-generation cephalosporins also show relatively high resistance levels, whereas amikacin has comparatively low resistance. Gentamicin and piperacillin with tazobactam have similar resistance distributions. The spread of values varies between antibiotic categories, although the differences may be difficult to assess visually because the resistance percentages vary considerably between categories.

These findings describe the observed distributions and do not establish the causes of differences in antibiotic resistance.


### Antibiotic Resistance Patterns by Year

A heatmap will be used to compare resistance percentages across the seven antibiotic categories and reporting years from 2021 to 2026. Colours will represent the mean resistance percentage for each antibiotic within each year, helping identify differences and patterns across categories over time.

The first and last years cover partial reporting periods, so comparisons involving these years will be interpreted with caution.


#### Code Explanation

The code groups the data by reporting year and antibiotic category, then calculates the mean resistance percentage for each combination. These values are displayed in a heatmap, where the colour intensity represents the mean resistance percentage. This makes it easier to compare resistance patterns across antibiotics and years.


```python
# Calculate mean resistance by year and antibiotic
yearly_resistance = df.pivot_table(
    index="stratum",
    columns="year",
    values="metric_value",
    aggfunc="mean"
)

# Create the heatmap
plt.figure(figsize=(12, 7))
sns.heatmap(
    yearly_resistance,
    annot=True,
    fmt=".1f",
    cmap="YlOrRd",
    linewidths=0.5,
    cbar_kws={"label": "Mean resistance (%)"}
)

# Add chart labels and formatting
plt.title("Mean E. coli Resistance by Antibiotic and Year in England")
plt.xlabel("Year")
plt.ylabel("Antibiotic")
plt.tight_layout()
plt.show()
```

#### Interpretation of Resistance Patterns by Year

The heatmap illustrates mean *E. coli* resistance percentages across seven antibiotic categories in England between 2021 and 2026. Co-amoxiclav consistently shows the highest resistance levels, while carbapenems show the lowest. Several other antibiotic categories, including ciprofloxacin, gentamicin and third-generation cephalosporins, show an overall increase in mean resistance over the period. In contrast, carbapenems remain relatively stable. These findings highlight differences in resistance patterns across antibiotic categories and over time but do not establish the causes of these changes. As 2021 and 2026 cover partial reporting periods, comparisons involving these years should be interpreted with caution.

**Key Takeaway:** Resistance levels differ considerably between antibiotic categories. Co-amoxiclav has persistently high resistance, while carbapenems remain consistently low. Several other categories show increasing mean resistance over the study period.
