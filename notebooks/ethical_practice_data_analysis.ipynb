### Checking for Surrounding Whitespace

This step checks whether text values contain unnecessary spaces at the beginning or end, which could cause inconsistencies during data analysis.


```python
# Check whether text columns contain leading or trailing whitespace.

text_columns = df.select_dtypes(include=["object", "string"]).columns

for column in text_columns:
    whitespace_count = (
        df[column].dropna()
        .ne(df[column].dropna().str.strip())
        .sum()
    )
    print(f"{column}: {whitespace_count} values with surrounding whitespace")
```

The check found no surrounding whitespace in the text columns, so no whitespace cleaning is required.

This analysis will summarise antibiotic resistance measurements for each antibiotic category using descriptive statistics. Comparing the mean, median, minimum and maximum values will help identify differences in resistance levels across the seven antibiotic categories.


```python
# Calculate descriptive statistics for each antibiotic category
antibiotic_summary = df.groupby("stratum")["metric_value"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

# Display the summary statistics
display(antibiotic_summary)
```

These are descriptive findings from the dataset and do not establish why resistance levels differ. Further analysis will examine how resistance patterns change over time.
